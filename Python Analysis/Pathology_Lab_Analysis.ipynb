## **<u>1. Database Connection & Ingestion (MySQL to Pandas)</u>**

```python
import pandas as pd
import seaborn as sns
import matplotlib .  pyplot as plt 
import numpy as np
import mysql . connector 
plt.style.use('dark_background')
```

```python
# Step 1: Establish Database Connection with Pathology_Lab_db

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_PASSWORD",
    database="pathology_lab_db",
    use_pure=True
)
print("Database Connected Succesfully")
cursor=connection.cursor()

```

```python
# Billing Table Extraction

cursor.execute("SELECT * FROM billing")
billing=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
billing_df = pd.DataFrame(billing, columns=columns)
print(billing)

# Company Table Extraction

cursor.execute("SELECT * FROM company")
company=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
company_df = pd.DataFrame(company, columns=columns)
print(company)

# date_dimension Table Extraction

cursor.execute("SELECT * FROM date_dimension")
date_dimension=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
date_dimension_df = pd.DataFrame(date_dimension, columns=columns)
print(date_dimension)

# department Table Extraction

cursor.execute("SELECT * FROM department")
department=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
department_df = pd.DataFrame(department, columns=columns)
print(department)

# doctor Table Extraction

cursor.execute("SELECT * FROM doctor")
doctor=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
doctor_df = pd.DataFrame(doctor, columns=columns)
print(doctor)

# interpretation Table Extraction

cursor.execute("SELECT * FROM interpretation")
interpretation=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
interpretation_df = pd.DataFrame(interpretation, columns=columns)
print(interpretation)

# patients Table Extraction

cursor.execute("SELECT * FROM patients")
patients=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
patients_df = pd.DataFrame(patients, columns=columns)
print(patients)

# sales Table Extraction

cursor.execute("SELECT * FROM sales")
sales=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
sales_df = pd.DataFrame(sales, columns=columns)
print(sales)

# test Table Extraction

cursor.execute("SELECT * FROM test")
test=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
test_df = pd.DataFrame(test, columns=columns)
print(test)

# test_discount Table Extraction

cursor.execute("SELECT * FROM test_discount")
test_discount=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
test_discount_df = pd.DataFrame(test_discount, columns=columns)
print(test_discount)

# test_procedure Table Extraction

cursor.execute("SELECT * FROM test_procedure")
test_procedure=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
test_procedure_df = pd.DataFrame(test_procedure, columns=columns)
print(test_procedure)

# test_record Table Extraction

cursor.execute("SELECT * FROM test_record")
test_record=cursor.fetchall()
if cursor.description is not None:
    columns = [i[0] for i in cursor.description]
test_record_df = pd.DataFrame(test_record, columns=columns)
print(test_record)
```

#### **STEP 2: Convert Date Columns to Datetime**
- Standardizing date columns to datetime format

```python
# billing Table
billing_df["bill_date"] = pd.to_datetime(billing_df["bill_date"],errors= 'coerce')
print("Billing:", billing_df["bill_date"].dtype)

# Date_Dimension Table
date_dimension_df["date"] = pd.to_datetime(date_dimension_df["date"],errors= 'coerce')
print("Date Dimension:", date_dimension_df["date"].dtype)

# Patients Table
patients_df["registration_date"] = pd.to_datetime(patients_df["registration_date"],errors= 'coerce')
print("Patients:", patients_df["registration_date"].dtype)

# Test_Discount Table
test_discount_df["valid_from"] = pd.to_datetime(test_discount_df["valid_from"],errors= 'coerce')
print("Discount From:", test_discount_df["valid_from"].dtype)
test_discount_df["valid_to"] = pd.to_datetime(test_discount_df["valid_to"],errors= 'coerce')
print("Discount To:", test_discount_df["valid_to"].dtype)

# Test_Record Table
test_record_df["test_date"] = pd.to_datetime(test_record_df["test_date"],errors= 'coerce',format='%Y-%m-%d')
print("Test Date:", test_record_df["test_date"].dtype)
test_record_df["report_date"] = pd.to_datetime(test_record_df["report_date"],errors= 'coerce')
print("Report Date:", test_record_df["report_date"].dtype)
```

#### **Turnaround Time (TAT) Distribution**

**Business Question:**  
How long does it take for the laboratory to generate test reports?

     Why does this matter?
 
Turnaround time is an important operational metric for laboratory performance. Understanding its distribution helps identify typical
reporting time and potential delays.

```python
# Calculate turnaround time for test records with a report date

tat_df = test_record_df[test_record_df["report_date"].notnull()].copy()
tat_df["tat_hours"] = (tat_df["report_date"] - tat_df["test_date"]).dt.total_seconds() / 3600
tat_df.head(5)
```

```python
# Visualize the distribution of turnaround time

plt.figure(figsize=(6,3))

sns.histplot(data=tat_df, x="tat_hours", bins=20, kde=True,color ="green",line_kws={"linewidth": 3, "color": "darkred"})

plt.title("Laboratory Test Turnaround Time Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Turnaround Time (Hours)", fontsize=12, fontweight="bold")
plt.ylabel("Number of Test Records", fontsize=12, fontweight="bold")

plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

#### **Turnaround Time by Department**

**Business Question:**  
How does laboratory turnaround time vary across departments?

     Why does this matter?
Comparing turnaround time across departments can help identify differences in operational workload and reporting efficiency.

```python
# Merge test records with test and department information

a1 = tat_df.merge(test_df, on="test_id")
a2= pd.merge(
    a1,
    department_df,
    left_on='dept_id',
    right_on='Dept_id',
    how='left')
print(a2)
```

```python
# Compare turnaround time across laboratory departments

plt.figure(figsize=(10, 5))
sns.boxplot(
    data=a2,
    x="Dept_name",
    y="tat_hours",
    hue="Dept_name",
    palette="magma",
    width=0.5,
    legend=True,
)

plt.title("Turnaround Time by Laboratory Department", fontsize=14, fontweight="bold")
plt.xlabel("Department", fontsize=12, fontweight="bold")
plt.ylabel("Turnaround Time (Hours)", fontsize=12, fontweight="bold")

plt.xticks(rotation=30, ha="right", fontsize=10)
plt.yticks(fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.legend(
    title="Dept_name",
    bbox_to_anchor=(1.02,1),
    loc="upper left",
    borderaxespad=0,
    frameon=True
)

plt.tight_layout()
plt.show()
```

**Business Question:**  
How are result statuses distributed across different diagnostic tests?

    Why does this matter? 
Test-wise result analysis helps identify differences in result patterns across commonly performed diagnostic tests and provides a deeper understanding of laboratory testing activity.

```python
# Analyze result status by diagnostic test

A1 = test_record_df.merge(test_df, on="test_id")
test_result_status = A1.groupby(["test_name", "result_status"])["record_id"].count().reset_index()
print(test_result_status.head(20))
```

#### **Result Status by Gender**

**Business Question:**  
How does the distribution of laboratory result status vary by gender?

    Why does this matter?
Comparing result status proportions across gender groups provides a demographic view of laboratory workflow outcomes within the dataset.

```python
# Analyze result status distribution by gender

l1 = test_record_df.merge(patients_df, on="patient_id")
gender_status = l1.groupby(["gender", "result_status"])["record_id"].count().reset_index()
gender_status_pivot = gender_status.pivot(index="gender", columns="result_status", values="record_id").fillna(0)
gender_status_percentage = gender_status_pivot.div(gender_status_pivot.sum(axis=1), axis=0) * 100
print(gender_status_percentage)
```

```python
# Visualize result status proportions by gender

gender_status_percentage.plot(kind="bar", stacked=True, figsize=(8, 5), colormap="Set2")

plt.title("Result Status Distribution by Gender", fontsize=14, fontweight="bold")
plt.xlabel("Gender", fontsize=12, fontweight="bold")
plt.ylabel("Percentage of Test Records", fontsize=12, fontweight="bold")

plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.legend(
    title="Result Status",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    borderaxespad=0,
)
plt.show()
```

```python
# Prepare test-level price, revenue, and discount data
A1 = sales_df.merge(test_df, on="test_id")
test_commercial = A1.groupby(["test_id", "test_name", "price"])["revenue"].sum().reset_index()

discount_summary = test_discount_df.groupby("test_id")["discount_percent"].mean().reset_index()

test_commercial = test_commercial.merge(discount_summary, on="test_id")
print(test_commercial.head(10))
```

```python
# Visualize correlations between price, discount, and revenue
plt.figure(figsize=(10, 4))
sns.heatmap(data=commercial_correlation, annot=True, fmt=".2f")
plt.title("Correlation Between Price, Discount and Revenue", fontsize=14, fontweight="bold")
plt.xlabel("Commercial Variables", fontsize=12, fontweight="bold")
plt.ylabel("Commercial Variables", fontsize=12, fontweight="bold")
plt.show()
```

- Discount percentage and revenue showed a weak positive correlation of approximately 0.29, indicating a limited positive relationship in the observed data.

- Test price and discount percentage showed a moderate positive correlation of approximately 0.41, indicating that higher-priced tests were associated with somewhat higher discount percentages in the dataset.

- The correlation results indicate that test price alone does not explain revenue variation, and test volume may also play an important role in overall revenue generation.

- These correlations represent associations in the dataset and do not establish causal relationships.
